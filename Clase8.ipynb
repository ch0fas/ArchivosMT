```python
import nltk 
from nltk.tokenize import word_tokenize, sent_tokenize
from nltk.corpus import stopwords
from nltk.stem import SnowballStemmer, wordnet, WordNetLemmatizer
```

```python
nltk.download("punkt")
nltk.download("punkt_tab")
```

```python
texto = "Esta oracion es una prueba. NLTK es una gran librería"

#Tokenizando por operaciones
oraciones = sent_tokenize(texto, language="spanish")

#Tokenizando por palabras
palabras = word_tokenize(texto, language="spanish")

print(oraciones)
print(palabras)
```

```python
nltk.download("stopwords")
```

```python
texto2 = "Esta es una oración de ejemplo para mostrar cómo eliminar las stopwords en español"

#Por palabras

palabras = word_tokenize(texto2, language="spanish")

#Filtrar stopwords
stop_words = set(stopwords.words("spanish"))

palabras_filtradas = [i for i in palabras if i.lower() not in stop_words]

print(palabras)
print(list(stop_words)[:15])
print(palabras_filtradas)
```

```python
# stemming en esp

stemmer_es = SnowballStemmer("spanish")

palabras = ['jugar','jugador','jugando','juegos','juguetes']

stems = [stemmer_es.stem(i) for i in palabras]

print(palabras)
print(stems)
```

```python
lemmatizer = WordNetLemmatizer()

palabras = ['running','runs','ran','radio']

lemmas = [lemmatizer.lemmatize(i, pos='v') for i in palabras]

print(palabras)
print(lemmas)
```

```python
# pip install -U spacy
# python -m spacy download en_core_web_sm
import spacy

# Load English tokenizer, tagger, parser and NER
nlp = spacy.load("en_core_web_sm")

# Process whole documents
text = ("When Sebastian Thrun started working on self-driving cars at "
        "Google in 2007, few people outside of the company took him "
        "seriously. “I can tell you very senior CEOs of major American "
        "car companies would shake my hand and turn away because I wasn’t "
        "worth talking to,” said Thrun, in an interview with Recode earlier "
        "this week.")
doc = nlp(text)

# Analyze syntax
print("Noun phrases:", [chunk.text for chunk in doc.noun_chunks])
print("Verbs:", [token.lemma_ for token in doc if token.pos_ == "VERB"])

# Find named entities, phrases and concepts
for entity in doc.ents:
    print(entity.text, entity.label_)
```
